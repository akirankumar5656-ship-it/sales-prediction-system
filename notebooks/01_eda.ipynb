# 01 — Exploratory Data Analysis

Sales Prediction System — EDA notebook.

This notebook calls the reusable plotting functions in `src/eda.py` so the exact same code that generates `reports/figures/` here can also be invoked from the training pipeline or a Flask analytics route. Nothing is duplicated — this notebook is a thin, visual wrapper around that module.

```python
import sys, os
sys.path.append(os.path.abspath('..'))  # so `src` is importable from notebooks/

%matplotlib inline
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['figure.dpi'] = 100

from src.eda import (
    load_data, CONFIG,
    plot_sales_distribution, plot_sales_over_time, plot_sales_by_product,
    plot_sales_by_category, plot_sales_by_region, plot_correlation_heatmap,
    plot_quantity_vs_sales, plot_price_vs_sales,
)
```

## Load data

```python
plot_sales_distribution(df)
plt.show()
```

## 2. Sales Over Time
Monthly aggregated sales trend — reveals seasonality and growth/decline patterns.

```python
plot_sales_over_time(df)
plt.show()
```

## 3. Sales by Product

```python
plot_sales_by_product(df)
plt.show()
```

## 4. Sales by Category
Total sales grouped by product category.

```python
plot_sales_by_category(df)
plt.show()
```

## 5. Sales by Region
Total sales grouped by region.

```python
plot_sales_by_region(df)
plt.show()
```

## 6. Correlation Heatmap
Correlation across all numeric columns — helps spot multicollinearity and which features are likely predictive of Sales.

```python
plot_correlation_heatmap(df)
plt.show()
```

## 7. Quantity vs Sales
Scatter + regression line — does higher order volume drive higher revenue?

```python
plot_quantity_vs_sales(df)
plt.show()
```

## 8. Price vs Sales
